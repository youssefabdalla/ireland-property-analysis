```python
from dataclasses import dataclass
```

```python
@dataclass
class card:
    rank: str
    suit: str
```

```python
card1 = card("A", "spades")
print(card1.rank)
```

```python
# type annotation
@dataclass
class Position:
    name: str
    lon: float
    lat: float
```

```python
pos = Position("Oslo", 10.8, 59.9)
print(pos)
```
